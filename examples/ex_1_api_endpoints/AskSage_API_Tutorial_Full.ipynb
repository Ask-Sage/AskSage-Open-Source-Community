## 3. Authentication -- Get Token with API Key

**`POST /user/get-token-with-api-key`** -- Exchange your API key for a 24-hour access token.

Ask Sage uses the `x-access-tokens` header for authentication. You can use either:

- **Static API key** -- use directly as the token value (no expiry)
- **24-hour access token** -- exchange your API key for a short-lived token via this endpoint

| Parameter | Type | Required | Description |
|-----------|------|----------|-------------|
| `email` | string | Yes | Your Ask Sage account email |
| `api_key` | string | Yes | Your static API key |

### Raw REST

```python
# Get a 24-hour access token
auth_response = requests.post(
    f"{USER_BASE_URL}/get-token-with-api-key",
    json={"email": EMAIL, "api_key": API_KEY}
)

auth_data = auth_response.json()
print(f"Status: {auth_response.status_code}")
print(f"Response: {auth_data.get('response', 'N/A')}")

# Store the token and build reusable headers
TOKEN = auth_data.get("access_token", API_KEY)  # fallback to static key
HEADERS = {
    "x-access-tokens": TOKEN,
    "Content-Type": "application/json"
}

print(f"Token obtained: {'Yes' if 'access_token' in auth_data else 'Using static API key'}")
```

### asksageclient

The client handles token management internally.

```python
from asksageclient import AskSageClient

client = AskSageClient(email=EMAIL, api_key=API_KEY)
print("Client initialized successfully")
```

## 4. Get User Logins

**`POST /user/get-user-logins`** -- Retrieve your recent login history. Returns timestamps and IP addresses.

| Parameter | Type | Required | Description |
|-----------|------|----------|-------------|
| `limit` | integer | No | Max login records to return (1-100, default: 5) |

### Raw REST

```python
resp = requests.post(
    f"{USER_BASE_URL}/get-user-logins",
    headers=HEADERS,
    json={"limit": 5}
)

logins_data = resp.json()
print(f"Status: {resp.status_code}")

if HAS_PANDAS and "response" in logins_data and isinstance(logins_data["response"], list):
    df = pd.DataFrame(logins_data["response"])
    if not df.empty:
        display(df.head(5))
    else:
        print("No login records found.")
else:
    print(json.dumps(logins_data, indent=2))
```

### Raw REST

```python
resp = requests.post(
    f"{USER_BASE_URL}/get-user-logs",
    headers=HEADERS,
    json={}
)

logs_data = resp.json()
print(f"Status: {resp.status_code}")

if HAS_PANDAS and "response" in logs_data and isinstance(logs_data["response"], list):
    df = pd.DataFrame(logs_data["response"])
    if not df.empty:
        cols = [c for c in ["date_time", "prompt", "completion"] if c in df.columns]
        display(df[cols].head(10))
    else:
        print("No logs found.")
else:
    print(json.dumps(logs_data, indent=2))
```

## 21. Query

**`POST /server/query`** -- The main endpoint for generating AI completions.

**Key parameters:**

| Parameter | Type | Required | Description |
|-----------|------|----------|-------------|
| `message` | string/array | Yes | Your query (string or conversation array) |
| `model` | string | No | AI model to use (default: `openai_gpt`) |
| `persona` | integer | No | Persona ID (default: 1) |
| `dataset` | string/array | No | Dataset(s) to query -- `"all"`, `"none"`, or specific names |
| `temperature` | number | No | Randomness 0.0-1.0 (default: 0.0) |
| `limit_references` | integer | No | Max knowledge base references (0 = no embeddings) |
| `live` | integer | No | Web search: 0=off, 1=Google, 2=Google+crawl |
| `system_prompt` | string | No | Custom system prompt (overrides default) |
| `tools` | array | No | Tool/function definitions (format depends on model) |
| `usage` | boolean | No | Include usage statistics in response (default: false) |

### Basic Query -- Raw REST

```python
resp = requests.post(
    f"{SERVER_BASE_URL}/query",
    headers=HEADERS,
    json={
        "message": "What is Ask Sage?",
        "model": "gpt-4o-mini",
        "temperature": 0.0
    }
)

query_data = resp.json()
print(f"Status: {resp.status_code}\n")
print("AI Response:")
print(query_data.get("message", "No response"))
```

### Basic Query -- asksageclient

```python
result = client.query("What is Ask Sage?", model="gpt-4o-mini")
print(result.get("message", "No response"))
```

### Advanced Query -- Raw REST

Using persona, dataset, live search, and custom system prompt.

```python
resp = requests.post(
    f"{SERVER_BASE_URL}/query",
    headers=HEADERS,
    json={
        "message": "What are the latest cybersecurity threats?",
        "model": "gpt-4o-mini",
        "persona": 1,
        "dataset": "none",
        "temperature": 0.7,
        "limit_references": 3,
        "live": 1,
        "system_prompt": "You are a cybersecurity analyst. Be concise and specific."
    }
)

adv_data = resp.json()
print(f"Status: {resp.status_code}\n")
print("AI Response:")
print(adv_data.get("message", "No response"))

refs = adv_data.get("references", [])
if refs:
    print(f"\nReferences ({len(refs)}):")
    for ref in refs[:3]:
        print(f"  - {ref}")
```

### Advanced Query -- asksageclient

```python
result = client.query(
    message="What are the latest cybersecurity threats?",
    model="gpt-4o-mini",
    persona=1,
    dataset="none",
    temperature=0.7,
    limit_references=3,
    live=1,
    system_prompt="You are a cybersecurity analyst. Be concise and specific."
)

print(result.get("message", "No response"))
```

## 53. OpenAI -- Chat Completions

**`POST /server/openai/v1/chat/completions`** -- Chat completions endpoint following the OpenAI API specification. Supports multi-turn conversations, function calling, and all standard OpenAI parameters.

| Parameter | Type | Required | Description |
|-----------|------|----------|-------------|
| `model` | string | Yes | AI model to use (e.g., `"gpt-4o-mini"`, `"claude-35-sonnet"`) |
| `messages` | array | Yes | Array of message objects with `role` and `content` |
| `temperature` | number | No | Randomness 0.0-2.0 (default: 1.0) |
| `max_tokens` | integer | No | Maximum tokens to generate |
| `top_p` | number | No | Nucleus sampling 0.0-1.0 (default: 1.0) |
| `frequency_penalty` | number | No | Reduce repeated tokens, -2.0 to 2.0 (default: 0.0) |
| `presence_penalty` | number | No | Encourage new topics, -2.0 to 2.0 (default: 0.0) |
| `tools` | array | No | Tool/function definitions for function calling |
| `tool_choice` | string | No | Controls tool usage: `"none"`, `"auto"`, or specific tool |

### Basic Chat Completion

```python
OPENAI_HEADERS_JSON = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

resp = requests.post(
    f"{OPENAI_BASE_URL}/chat/completions",
    headers=OPENAI_HEADERS_JSON,
    json={
        "model": "gpt-4o-mini",
        "messages": [
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": "What is Ask Sage?"}
        ]
    }
)

chat_data = resp.json()
print(f"Status: {resp.status_code}\n")

if "choices" in chat_data:
    print("AI Response:")
    print(chat_data["choices"][0]["message"]["content"])
    usage = chat_data.get("usage", {})
    print(f"\nTokens -- prompt: {usage.get('prompt_tokens')}, completion: {usage.get('completion_tokens')}, total: {usage.get('total_tokens')}")
else:
    print(json.dumps(chat_data, indent=2))
```

## 54. Tips & Best Practices

**Token Management**
- Use **static API keys** for development and testing
- Use **24-hour access tokens** for production applications -- they limit exposure if compromised

**Error Handling**
- Always check `response.status_code` before parsing JSON
- Handle `401` responses by refreshing your token
- Wrap API calls in try/except for network errors

**Performance**
- Set `limit_references=0` if you don't need RAG context (faster responses)
- Use `live=0` unless you specifically need web search results
- Use `temperature=0.0` for deterministic, factual responses

**Datasets**
- Use descriptive `context` values when training -- they help the AI find relevant content
- Dataset names must be alphanumeric (no spaces) -- create via the **User API** `/add-dataset`
- Train content into datasets via the **Server API** `/train`, `/train-with-file`, or `/train-with-array`
- Monitor teach token usage with `GET /server/count-monthly-teach-tokens`
- Monitor query token usage with `GET /server/count-monthly-tokens`

**MCP Servers**
- Use `/list-mcp-whitelisted-servers` to see which servers are approved for your tenant
- Use `/list-mcp-tools` to discover available tools before building workflows
- Deleted MCP servers are soft-deleted and can potentially be restored

**Agent Builder**
- Always use `streaming=False` for simple integrations; use `streaming=True` for real-time UI updates
- Pass `variables` to customize agent behavior at runtime
- Use `conversation_history` for multi-turn agent interactions

**OpenAI Compatibility**
- Use `Bearer` token in the `Authorization` header (not `x-access-tokens`)
- The endpoint follows the exact OpenAI API spec -- existing OpenAI code works with minimal changes
- Just swap the base URL from `https://api.openai.com/v1` to `https://api.asksage.ai/server/openai/v1`

**API Structure Reminder**
- **User API** (`api.asksage.ai/user/`) -- Authentication, dataset creation, user management, chat history, API keys
- **Server API** (`api.asksage.ai/server/`) -- AI queries, training, models, plugins, MCP, agents, utilities
- **OpenAI Compat** (`api.asksage.ai/server/openai/v1/`) -- Drop-in OpenAI-compatible chat completions
